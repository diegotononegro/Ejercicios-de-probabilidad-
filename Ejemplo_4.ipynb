#Ejemplo 4

```python
from itertools import product

S = list(product(range(10), repeat=3))
print("Espacio Muestral:", S)

A = [x for x in S if x[0] == x[1] == x[2]]
print("Casos favorables:", A[:10])  # solo muestra algunos

print("Probabilidad:", len(A)/len(S))
```

##Ejemplo 5 parte clásica

Lanzar 3 monedas y que exactamente 2 sean aguila

El espacio muestral es

$$Ω={"A","S"}×{"A","S"}×{"A","S"}$$

Por lo tanto el total de eventos son

$$Ω=2×2×2=8$$

Buscamos todos los elementos que tiene 2 caras, aplicando la fórmula tenemos

$$|A|=(\frac{3}{2})=\frac{3!}{2!(2−3)!}=\frac{3⋅2⋅1}{2⋅1⋅1!}=\frac{6}{2⋅1}=\frac 62 = 3$$

```python
from itertools import product

S = list(product(["A","S"], repeat=3))
print("Espacio Muestral:", S)

A = [x for x in S if x.count("A") == 2]
print("Casos favorables:", A)

print("Probabilidad:", len(A)/len(S))
```
