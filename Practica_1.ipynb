Lanzamiento de una moneda


```python
import random

moneda_lados = ["aguila", "sol"]

def lanzar_moneda():
    return random.choice(moneda_lados)

```

```python
lanzar_moneda()
```

Lanzamiento de un dado

```python
import random

dado_lados = ["uno", "dos","tres","cuatro","cinco","seis"]

def lanzar_dado():
    return random.choice(dado_lados)

lanzar_dado()
```
